# PRAKTIKUM 04 - FUNGSI DENGAN *ARGS DAN **KWARGS

```python
def jumlahkan(*args):
    return sum(args)
def biodata(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print("Total:", jumlahkan(1, 2, 3, 4, 5))
biodata(Nama="Budi", Umur=20, Prodi="TRK")
```

# **PRAKTIKUM 07 - FUNGSI DECORATOR**

```python
def decorator_sapa(fungsi):
    def wrapper():
        print("Sebelum fungsi dijalankan...")
        fungsi()
        print("Setelah fungsi dijalankan...")
    return wrapper

@decorator_sapa
def sapa():
    print("Halo, Mahasiswa Polines!")

sapa()
```

# **PENUGASAN 05 - FUNGSI DENGAN KWARGS**

# **PENUGASAN 08 - FUNGSI DECORATOR**

```python
def logger(fungsi):
    def wrapper():
        print("Memulai fungsi...")
        fungsi()
        print("Fungsi selesai dijalankan.")
    return wrapper

@logger
def halo():
    print("Halo, Selamat belajar Python.")

halo()
```

# **PENUGASAN 09 - STUDI KASUS NYATA - PROGRAM NILAI**

```python
# a. Fungsi input_nilai untuk menerima nilai mahasiswa
def input_nilai():
    nilai = []
    banyak = int(input("Masukkan jumlah nilai: "))
    for i in range(banyak):
        n = float(input(f"Nilai ke-{i+1}: "))
        nilai.append(n)
    return nilai

# b. Fungsi menghitung rata-rata
def hitung_rata(data):
    if len(data) == 0:
        return 0
    return sum(data) / len(data)

# c. Fungsi menentukan status kelulusan
def status_kelulusan(rata):
    if rata >= 60:
        return "Lulus"
    else:
        return "Tidak Lulus"

# d. Lambda menghitung nilai akhir (40% tugas, 60% ujian)
nilai_akhir = lambda tugas, ujian: (0.4 * tugas) + (0.6 * ujian)

# --- Program Utama ---
# (contoh penggunaan tanpa input untuk demonstrasi)
print("=== Program Nilai Mahasiswa ===")

# Contoh input manual tanpa input()
nilai_mahasiswa = [80, 75, 90]  # bisa diganti atau pakai input_nilai()

rata = hitung_rata(nilai_mahasiswa)
print("Nilai:", nilai_mahasiswa)
print("Rata-rata:", rata)
print("Status:", status_kelulusan(rata))

# Contoh hitung nilai akhir dengan lambda
tugas = 85
ujian = 70
print("Nilai akhir:", nilai_akhir(tugas, ujian))
```
